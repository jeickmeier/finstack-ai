# First agent

Goal: run one offline agent through `PythonModel` and `Agent.from_python`.

Trust: T2 trusted callback. Network: none. Mirrors `examples/python-minimal/python-callback/main.py`.


```python
from typing import Any

import finstack_ai


async def model(
    context: finstack_ai.CallbackContext, request: dict[str, Any]
) -> dict[str, Any]:
    assert context.kind == "model"
    assert request["model"] == "notebook-model"
    return {"text": "hello from the Rust-owned run loop", "completion_id": "notebook-1"}


agent = await finstack_ai.Agent.from_python(
    finstack_ai.PythonModel(
        model,
        component="notebook.model.first",
        provider="notebook",
        model="notebook-model",
    ),
    instruction="Reply with the scripted completion.",
)
result = await agent.run("hello")
print(result.text)
print(result.trace)
```
